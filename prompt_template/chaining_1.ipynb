## **Maintain chat history by chatbot**

```python
chat_history = [
    SystemMessage(content="your an intelligent bot and very helpfull")
]
```

```python
while True:
    user_input = input("User : ")
    chat_history.append(HumanMessage(content = user_input))
    if user_input == "exit":
        break
    result = groq_model.invoke(chat_history)
    chat_history.append(AIMessage(result.content))
    print("AI MESSAGE: ", result.content)
```

```python
print(chat_history)
```
